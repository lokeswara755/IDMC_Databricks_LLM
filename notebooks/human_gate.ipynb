```python
# ============================================================
# CELL 1: INITIALIZE HUMAN GATEWAY
# ============================================================

import json
from Human_GateWay.human_gateway import HumanGateway

REVIEW_DIR = (
    "/Volumes/idmc_poc/bronze/idmc_files/"
    "generated_pyspark/human_reviews"
)

gateway = HumanGateway(review_dir=REVIEW_DIR)

print("=" * 80)
print("HUMAN GATEWAY INITIALIZED")
print("=" * 80)

print("Review directory:", REVIEW_DIR)
print("Gateway ready!")
```

```python
# ============================================================
# CELL 2: DISPLAY ALL HUMAN GATEWAY REVIEWS
# ============================================================

all_reviews = gateway.list_reviews()

print("=" * 100)
print("ALL HUMAN GATEWAY REVIEWS")
print("=" * 100)

print("Total reviews:", len(all_reviews))

for review in all_reviews:

    decision = review.get("decision") or {}

    print("\nMapping:", review.get("mapping_name"))
    print("Review ID:", review.get("review_id"))
    print("Scope:", review.get("scope"))
    print("Status:", review.get("status"))
    print("Decision:", decision.get("decision"))
    print("Validations passed:", review.get("validations_passed"))
    print("Can execute:", review.get("can_execute"))
    print("Issues:")

    print(json.dumps(
        review.get("issues", []),
        indent=2,
        default=str
    ))

    print("-" * 100)
```
